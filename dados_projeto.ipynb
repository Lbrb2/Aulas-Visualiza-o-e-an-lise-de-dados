```python
import kagglehub
import pandas as pd
import matplotlib.pyplot as plt


url = kagglehub.dataset_download("algozee/teenager-menthal-healy")
df = pd.read_csv(f"{url}/Teen_Mental_Health_Dataset.csv", on_bad_lines= "skip")
df
```

##### Informações do DataFrame

```python
df.info()
df.describe(include = "all")
```

##### Criando faixas etárias



```python
df['age_group'] = pd.cut(df['age'],
                         bins=[12,15,17,20],
                         labels=['13-14','15-16','17-19'])
```

##### Média de ansiedade por faixa etária

```python
df.groupby('age_group')['anxiety_level'].mean().round(2)
```

##### Comparando ansiedade média entre faixas etárias

```python
media = df.groupby('age_group')['anxiety_level'].mean()

cores = ['#A8DADC','#FFD6A5', '#CDE7F0']

media.plot(kind='bar', color=cores)

plt.title('Idade vs Nível Médio de Ansiedade')
plt.xlabel('Faixa Etária')
plt.ylabel('Ansiedade Média')


plt.show()
```

##### Classificando o uso diário de redes sociais

```python
df['social_group'] = pd.cut(df['daily_social_media_hours'],
                           bins=[0,2,4,6,10],
                           labels=['baixo','moderado','alto','extremo'])
```

##### Quantas pessoas há em cada faixa de uso das redes sociais


```python
df['social_group'].value_counts()
```

##### Média de ansiedade por faixa de uso das redes sociais

```python
df.groupby('social_group')['anxiety_level'].mean().round(2)
```

##### Gráfico da ansiedade média por nível de uso das redes sociais

```python
media = df.groupby('social_group')['anxiety_level'].mean()

cores = ['#A8DADC', '#FFD6A5', '#BDB2FF', '#FFAFCC']


ax = media.plot(kind='bar', color=cores)

plt.title('Tempo de Redes Sociais vs Nível Médio de Ansiedade')
plt.xlabel('Nível de Uso')
plt.ylabel('Ansiedade Média')

# valores nas barras
for i, v in enumerate(media):
    ax.text(i, v + 0.02, f"{v:.2f}", ha='center')

plt.show()
```

#####  Média de ansiedade por faixa etária e nível de uso das redes sociais

```python
df.groupby(['age_group','social_group'])['anxiety_level'].mean().round(2)
```

##### Gráfico da ansiedade média por uso de redes sociais e faixa etária

```python
tabela = df.groupby(['social_group','age_group'])['anxiety_level'].mean().unstack()

tabela.plot(kind='bar', colormap='Pastel1')

plt.title('Idade + Uso de Redes Sociais vs Ansiedade')
plt.xlabel('Nível de Uso')
plt.ylabel('Ansiedade Média')

plt.show()
```

##### Soma de casos de depressão por nível de uso das redes sociais

```python
df.groupby('social_group')['depression_label'].sum()
```

##### Média de depressão por nível de uso das redes sociais

```python
df.groupby('social_group')['depression_label'].mean().round(3)
```

##### Taxa média de depressão por nível de uso das redes sociais
