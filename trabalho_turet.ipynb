### Importa bibliotecas, baixa o dataset do Kaggle e carrega os dados dos livros em um DataFrame


```python
import kagglehub
import pandas as pd
import plotly.express as px


url = kagglehub.dataset_download("jealousleopard/goodreadsbooks")

df = pd.read_csv(f"{url}/books.csv", on_bad_lines="skip")

df.head()
```

### Exibe as últimas linhas do DataFrame


```python
df.tail()
```

### Remove as colunas do DataFrame


```python
df = df.drop(columns=['isbn', 'isbn13'])
```

```python
df.columns
```

### Exibe informações resumidas sobre o DataFrame

```python
df.info()
```

### Verifica a quantidade de valores nulos em cada coluna do DataFrame


```python
df.isnull().sum()
```

### Mostra estatísticas descritivas das colunas numéricas do DataFrame


```python
df.describe()
```

### Remove os espaços das colunas e cria a coluna book_size


```python
df.rename(columns=lambda x: x.strip(), inplace=True)

df["book_size"] = pd.cut(
    df["num_pages"],
    bins=[0, 200, 400, df["num_pages"].max()],
    labels=["short", "medium", "long"])


df.head()
```

### Conta quantos livros existem em cada categoria de tamanho


```python
df["book_size"].value_counts()
```

### Calcula a média das avaliações por categoria de tamanho de livro


```python
df.groupby("book_size")["average_rating"].mean().round(2)

```

### Média de avaliação por tamanho de livro em gráfico de barras


```python

ratings_by_size = df.groupby("book_size", observed=True)["average_rating"].mean().round(2).reset_index()

fig = px.bar(
    ratings_by_size,
    x="book_size",
    y="average_rating",
    text="average_rating",
    color="book_size",
    color_discrete_map={"short": "skyblue", "medium": "orange", "long": "green"},
    title="Average Rating by Book Size"
)

fig.update_traces(texttemplate='%{text:.2f}', textposition='outside')
fig.update_layout(
    xaxis_title="Book Size",
    yaxis_title="Average Rating",
    yaxis=dict(range=[3.5, 4.2]),
    showlegend=False
)

fig.show()
```

###  Livros com média máxima exibidos por tamanho em gráfico de barras



```python
df_max_rating = df.query("average_rating == 5.0")


fig = px.bar(df_max_rating,
             x="book_size",
             y="average_rating",
             color="book_size",
             title="Distribution of Books with Maximum Rating by Size")

fig.show()
```

### Quantidade de livros por autor, título e idioma que receberam nota máxima

```python
df_max_rating[["authors","title","language_code","average_rating"]].value_counts()
```

### Verifica se os 10 autores com mais livros tiveram algum livro com nota máxima



```python
authors_with_most_books = df["authors"].value_counts().head(10)
for autor in authors_with_most_books.keys():
    if autor in df_max_rating["authors"]:
        print(autor)

else:
    print("None of the 10 authors with the most published books received a perfect score for any of their works.")
```

### Top 10 autores com mais livros

```python
df_authors = df["authors"].value_counts().head(10).reset_index()
df_authors
```

### Gráfico de barras mostrando a quantidade de livros por autor


```python
fig = px.bar(df_authors, x='count', y='authors', color='authors', title='Number of Books per Author')
fig.show()
```

### Gráfico de dispersão de avaliações vs. avaliações de texto por tamanho de livro


```python
fig = px.scatter(df, x='ratings_count', y='text_reviews_count', color='book_size', title='Ratings vs Text Reviews')
fig.show()
```

### Gráfico de caixa da distribuição das avaliações médias nos 5 idiomas mais comuns


```python
top_5_languages = df.groupby('language_code')['average_rating'].count().sort_values(ascending=False).head(5).index.tolist()

df_filtered = df[df['language_code'].isin(top_5_languages)]
fig = px.box(df_filtered, x="language_code", y="average_rating", title="Distribution of Average Ratings in the 5 Most Common Languages", color="language_code", points="all")
fig.show()
```

###  Barras da distribuição do tamanho dos livros nos 5 idiomas mais comuns



```python
top_5_languages = df.groupby('language_code', observed=True)['book_size'].count().sort_values(ascending=False).head(5).index.tolist()

df_filtered = df[df['language_code'].isin(top_5_languages)]
df_counts = df_filtered.groupby(['language_code', 'book_size'], observed=True).size().reset_index(name='Count')
fig = px.bar(df_counts, x="language_code", y="Count", color="book_size", title="Distribution of Book Sizes in the 5 Most Common Languages", barmode="group")

fig.show()
```

### Gráfico de dispersão da popularidade vs. qualidade dos livros por tamanho


```python
fig = px.scatter(df,
                 x='ratings_count',
                 y='average_rating',
                 size='text_reviews_count',
                 color='book_size',
                 title='Relationship between Popularity and Quality')
fig.show()
```
